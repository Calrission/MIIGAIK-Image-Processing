```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread('static/3/coins_1_1_2_5.jpg')
img_grey = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_grey_blur = cv2.GaussianBlur(img_grey, (5, 5), 0)
img_edges = cv2.Canny(img_grey_blur, 50, 150, apertureSize=3)
```

```python
plt.figure(figsize=(15, 5))
plt.subplot(1, 4, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Оригинал')

plt.subplot(1, 4, 2)
plt.imshow(img_grey, cmap='gray')
plt.axis('off')
plt.title('Оттенки серого')

plt.subplot(1, 4, 3)
plt.imshow(img_grey_blur, cmap='gray')
plt.axis('off')
plt.title('Размытие Гаусса')

plt.subplot(1, 4, 4)
plt.imshow(img_edges, cmap='gray')
plt.axis('off')
plt.title('Края Кэнни')
plt.show()
```

```python
dp = 1.2               # Разрешение обратного преобразования (чем больше, тем меньше кругов)
minDist = 30          # Минимальное расстояние между центрами обнаруженных кругов
param1 = 80           # Верхний порог для оператора Кэнни (нижний в два раза меньше)
param2 = 100            # Порог аккумулятора (чем меньше, тем больше ложных кругов)
minRadius = 40         # Минимальный радиус монеты в пикселях
maxRadius = 100        # Максимальный радиус монеты в пикселях

circles = cv2.HoughCircles(
    img_grey_blur,
    cv2.HOUGH_GRADIENT,
    dp=dp,
    minDist=minDist,
    param1=param1,
    param2=param2,
    minRadius=minRadius,
    maxRadius=maxRadius
)

if circles is not None:
    circles = np.round(circles[0, :]).astype("int")
    vis = img.copy()
    for i, (x, y, r) in enumerate(circles):
        cv2.circle(vis, (x, y), r, (0, 255, 0), 2)

    plt.figure(figsize=(12, 12))
    plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.title(f'Найдено монет: {len(circles)}')
    plt.show()
else:
    circles = []
    print("Монеты не обнаружены. Попробуйте изменить параметры преобразования Хафа.")
```

```python
def detect_lines_in_circle(image, cx, cy, r, canny_low=50, canny_high=150,
                           hough_thresh=20, min_line_len=10, max_line_gap=5):
    h, w = image.shape[:2]

    mask = np.zeros((h, w), dtype=np.uint8)
    cv2.circle(mask, (cx, cy), r, 255, -1)

    x1 = max(cx - r, 0)
    y1 = max(cy - r, 0)
    x2 = min(cx + r, w)
    y2 = min(cy + r, h)
    roi = image[y1:y2, x1:x2]
    roi_mask = mask[y1:y2, x1:x2]

    roi_gray = cv2.cvtColor(roi, cv2.COLOR_BGR2GRAY) if roi.ndim == 3 else roi
    roi_blur = cv2.GaussianBlur(roi_gray, (3, 3), 0)
    roi_edges = cv2.Canny(roi_blur, canny_low, canny_high)
    roi_edges = cv2.bitwise_and(roi_edges, roi_edges, mask=roi_mask)

    lines = cv2.HoughLinesP(
        roi_edges,
        rho=1,
        theta=np.pi / 180,
        threshold=hough_thresh,
        minLineLength=min_line_len,
        maxLineGap=max_line_gap
    )

    result_lines = []
    if lines is not None:
        lines = lines.reshape(-1, 4)
        for lx1, ly1, lx2, ly2 in lines:
            result_lines.append([int(lx1) + x1, int(ly1) + y1,
                                 int(lx2) + x1, int(ly2) + y1])

    # Полная карта краёв (для визуализации)
    full_edges = np.zeros((h, w), dtype=np.uint8)
    full_edges[y1:y2, x1:x2] = roi_edges

    return result_lines, full_edges
```

```python
def count_lines(lines):
    vertical_count = 0
    horizontal_count = 0
    for (x1, y1, x2, y2) in lines:
        dx, dy = x2 - x1, y2 - y1
        angle = np.degrees(np.arctan2(dy, dx)) % 180
        if angle < 20 or angle > 160:
            horizontal_count += 1
        elif 70 < angle < 110:
            vertical_count += 1
    return vertical_count, horizontal_count
```

```python
result = img.copy()
all_edges = np.zeros(img.shape[:2], dtype=np.uint8)
for i, (cx, cy, r) in enumerate(circles):
    lines, edges = detect_lines_in_circle(
        img, cx, cy, r,
    )
    cv2.circle(result, (cx, cy), r, (0, 255, 0), 2)

    for (x1, y1, x2, y2) in lines:
        cv2.line(result, (x1, y1), (x2, y2), (0, 0, 255), 2)
    all_edges = cv2.bitwise_or(all_edges, edges)

    vertical_count, horizontal_count = count_lines(lines)
    print(f"Монета {i} радиусом {r}: горизонтальных={horizontal_count}, вертикальных={vertical_count}")

plt.figure(figsize=(15, 5))
plt.imshow(all_edges, cmap='gray')
plt.axis('off')
plt.title('Края внутри обнаруженных монет')
plt.show()
```

```python
COINS = [
    #  ном R  V  H
    (1, 87, 56, 43),
    (2, 95, 56, 74),
    (5, 100,56, 64),
    (0.5, 77, 51, 65),
    (0.1, 74, 38, 80),
    (0.05, 79, 29, 89),
    (0.01, 62, 6,  46),
]

TOLERANCE = {
    "radius":     2,
    "vertical":   4,
    "horizontal": 4,
}


def _vote(value, refs, tol):
    in_range = [(nom, abs(value - ref)) for nom, ref in refs if abs(value - ref) <= tol]
    if in_range:
        return min(in_range, key=lambda x: (x[1], x[0]))[0]
    return min(refs, key=lambda kv: abs(value - kv[1]))[0]


def classify_coin(radius, vertical_count, horizontal_count):
    votes_by_criterion = {
        "radius":     _vote(radius,
                            [(n, r) for n, r, v, h in COINS],
                            TOLERANCE["radius"]),
        "vertical":   _vote(vertical_count,
                            [(n, v) for n, r, v, h in COINS],
                            TOLERANCE["vertical"]),
        "horizontal": _vote(horizontal_count,
                            [(n, h) for n, r, v, h in COINS],
                            TOLERANCE["horizontal"]),
    }

    vote_counts = {nom: 0 for nom, *_ in COINS}
    for nom in votes_by_criterion.values():
        vote_counts[nom] += 1

    if max(vote_counts.values()) == 1:
        return votes_by_criterion["radius"], votes_by_criterion

    max_count = max(vote_counts.values())
    tied = [nom for nom, cnt in vote_counts.items() if cnt == max_count]

    if len(tied) == 1:
        winner = tied[0]
    else:
        winner = (votes_by_criterion["radius"]
                  if votes_by_criterion["radius"] in tied
                  else min(tied))

    return winner, votes_by_criterion
```

```python
result = img.copy()
all_edges = np.zeros(img.shape[:2], dtype=np.uint8)

coins = []
total_sum = 0

circles_with_nominals = []

for i, (cx, cy, r) in enumerate(circles):
    lines, edges = detect_lines_in_circle(img, cx, cy, r)
    vertical_count, horizontal_count = count_lines(lines)
    nominal, votes = classify_coin(r, vertical_count, horizontal_count)
    coins.append((cx, cy, r, nominal))
    total_sum += nominal

    cv2.circle(result, (cx, cy), r, (0, 255, 0), 2)
    for (x1, y1, x2, y2) in lines:
        cv2.line(result, (x1, y1), (x2, y2), (0, 0, 255), 2)

    circles_with_nominals.append((cx, cy, r, nominal))

    all_edges = cv2.bitwise_or(all_edges, edges)

    print(f"Монета {i}: r={r}, V={vertical_count}, H={horizontal_count} → "
          f"R={votes['radius']}, V={votes['vertical']}, H={votes['horizontal']} → "
          f"{nominal} руб.")

for i, (cx, cy, r, nominal) in enumerate(circles_with_nominals):
    cv2.putText(result, f"{nominal} руб.", (cx - 30, cy - 10),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 0, 0), 2)

plt.figure(figsize=(15, 8))
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title(f"Обнаружено монет: {len(coins)}. Общая сумма: {total_sum} руб.")
plt.show()
```
